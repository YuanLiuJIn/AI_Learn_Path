- online / offline：学习时还能不能继续和环境交互、采新数据。
    - offline RL: $D = \{(s,a,r,s')\} \sim \mu$
    - 数据来自某个行为策略 $\mu$，训练时不能再用当前 $\pi$ 去环境里采样。因此只要你不断改进 $\pi$，就会面对 $\pi$ 和 $\mu$ 的分布偏移。
- on-policy / off-policy：用于更新的数据，是否来自“当前正在优化的目标策略” $\pi_\theta$。
    - $\text{on-policy} \iff d^\pi(s,a) \approx d^{\text{data}}(s,a)$
    - 训练数据的状态-动作分布来自当前目标策略 $\pi$。PPO 这类算法每轮都用当前或非常近的策略采样，所以是 online on-policy。SAC/DQN 有 replay buffer，里面混着过去很多策略的数据，所以是 online off-policy。
- 组合
    - online + on-policy
    - online + off-policy
    - offline + off-policy
        - 只能用固定数据集，数据来自行为策略 $\mu$，学习目标策略 $\pi$

- offline: 不能与环境交互，只能通过一个静态的历史数据集（Dataset）来学习策略。
    - 静态的 Dataset（比如包含 millions 条 (state, action, reward) 的历史记录）。环境是死的（或者是历史快照）。
    - 典型算法
        - td3 + bc（行为克隆）：$Loss = L_{TD3} + \lambda \|\pi(s) - a_{data}\|^2$
        - dpo
