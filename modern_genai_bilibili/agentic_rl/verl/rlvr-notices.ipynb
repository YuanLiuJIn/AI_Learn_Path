即使你精心设计了区分度，GRPO 的归一化那一步会把它重新抹掉。 论文的例子极其清楚 —— 两个 reward、两条 rollout 时，reward 组合 (0,1)、(0,2)、(1,2) 三种情况在求和后做 group z-score，全部塌缩成同一个 advantage $(-0.7071, +0.7071)$。信息全丢。这叫 reward signal collapse。

```python
import torch

for rewards in ([0., 1.], [0., 2.], [1., 2.]):
    r = torch.tensor(rewards)
    adv = (r - r.mean()) / r.std(correction=1)
    print(rewards, "->", adv.tolist())
```
