### detach

- 调用 tensor.detach() 的本质操作，是直接在计算图上用“无形的剪刀”剪断了连接该张量与生成它的操作节点之间的边。它从此成为了一个毫无历史包袱的新叶子节点，梯度反向回传时将在此处戛然而止。

```python
import torch

x = torch.tensor(2.0, requires_grad=True)
w = torch.tensor(3.0, requires_grad=True)

y = x * w
z = y + 1
loss = z ** 2
```

```python
loss.backward()
```

```python
x.grad, w.grad
```

```python
x = torch.tensor(2.0, requires_grad=True)
w = torch.tensor(3.0, requires_grad=True)

y = x * w
y_detach = y.detach()
z = y_detach + 1
loss = z ** 2
loss.backward()
```

- `.requires_grad_()`，它会成为一枚新的叶子变量，新的图从这里重新开始，旧图中的 x / w 不再接收梯度。

```python
x = torch.tensor(2.0, requires_grad=True)
w = torch.tensor(3.0, requires_grad=True)

y = x * w
y_detach = y.detach().requires_grad_()
z = y_detach + 1
loss = z ** 2
loss.backward()
```

```python
x.grad, w.grad
```
