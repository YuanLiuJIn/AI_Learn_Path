### entropy of sequence modeling

$$
\ell=\mathbb E_{y\sim p_\theta(y|x)}[-\log p_\theta(y|x)]
$$
- 对图像分类模型来说,输出类别有限,entropy 可以直接计算。语言模型或语音识别模型输出的 是序列,所有可能序列几乎无法穷举,所以完整 entropy.
    - $E_{y\sim p_\theta(y|x)}$: 对模型可能生成的所有序列取期望。
- 实际实现通常先从模型采样一个序列 $y = (y_1, \cdots, y_T)$。在生成每个 token 时, 模型都会给出下一个 token 的概率分布。单个 token 的分布有限, 可计算 entropy;于是把一路上每个 token 的  entropy 加起来, 得到可计算的 proxy loss:

$$
\tilde \ell(y)=\sum_{i=1}^TH(y_{1:i})
$$
- $\tilde \ell(y)$: 给定采样序列 $y$ 时实际可计算的代理损失；
