- ChatML 把结构与内容放在**同一个命名空间**里，靠"用户大概不会打出 `<|im_end|>`"这个约定支撑；XTML 把它们放进两个不相交的命名空间。

### xml -> xtml (eXtensible **Token** Markup Language，可扩展词元标记语言)

普通 XML 使用：
```
<message role="assistant">
  <think>...</think>
  <response>...</response>
</message>
```
XTML 把尖括号边界替换为保留控制词元：
```
<|open|>message role="assistant"<|sep|>
  <|open|>think<|sep|>...<|close|>think<|sep|>
  <|open|>response<|sep|>...<|close|>response<|sep|>
<|close|>message<|sep|><|end_of_msg|>
```

- `<|open|>`：开始一个结构。
- `<|sep|>`：标签头结束、正文开始。
- `<|close|>`：结构结束。
- `<|end_of_msg|>`：整条消息完成，也是生成停止点。
