### mc vs. td

$$
G_t = r_t + \gamma r_{t+1} + \gamma^2 r_{t+2} + \cdots
$$
- TD($\lambda$) 是一族 return 估计器；$\lambda=0$ 时完全依赖一步 bootstrap，$\lambda=1$ 时接近完整 Monte Carlo return。
    - $\hat G_t^{TD(0)} = r_t + \gamma V(s_{t+1})$
    - $\hat G_t^{TD(1)} =r_t + \gamma r_{t+1} + \gamma^2 r_{t+2} + \cdots + \gamma^{T-t-1}r_{T-1}$
- TD(0) 是“先用一步真实经验，再问 value network 未来值多少”；TD(1) 是“把后面真实发生的 reward 全部等完、加完”。

| 方法 | 公式直觉 | 是否 bootstrap | 需要等 episode 结束吗 | 方差 | 偏差 |
|---|---|---:|---:|---:|---:|
| TD(0) | $r_t + \gamma V(s_{t+1})$ | 是 | 否 | 低 | 较高 |
| TD(1) / MC | $r_t + \gamma r_{t+1} + \gamma^2 r_{t+2} + \cdots$ | 否，若完整 episode | 是 | 高 | 低 |
| TD($\lambda$) | 多个 $n$-step return 的 $\lambda$-加权平均 | 部分是 | 不一定 | 中间 | 中间 |

### ppo

> actor-critic + GAE + clipped policy gradient

- $\delta_t = r_t + \gamma V(s_{t+1}) - V(s_t)$
    - $\delta_t=r_t+\gamma m_t V_\phi(s_{t+1})-V_\phi(s_t)$
        - $m_t$ 对应 nextnonterminal，如果下一步已经 terminal，就是 0；否则是 1
    - `delta = rewards[t] + args.gamma * nextvalues * nextnonterminal - values[t]`
- advantages：$\hat A_t=\delta_t+\gamma \lambda m_t \hat A_{t+1}$
- returns：$\hat R_t=\hat A_t + V_\phi(s_t)$
    - $A(s_t,a_t) = Q(s_t,a_t) - V(s_t)$
    - $Q(s_t,a_t) \approx \hat R_t = \hat A_t + V(s_t)$
