$$
P(Z|X) = \frac{P(X|Z)P(Z)}{P(X)}= \frac{P(X|Z)P(Z)}{\int P(X,Z)dZ} =\frac{P(X|Z)P(Z)}{\int P(X|Z)P(Z) dZ}
$$

### 条件概率与积分

- $p(a,b|c)=p(a|b,c)p(b|c)$
    - 条件概率的链式法则（Conditional Chain Rule of Probability）

- 输入上下文 $\mathbf{C}$ (Context)
- 当前隐状态 $Z_0$ (Current Latent State)
- 目标隐状态 $Z_g$ (Goal Latent State)
- 动作序列 $\mathbf{A}$ (Action Sequence) 

$$
P(\mathbf{A}|\mathbf{C}) = \sum_{Z_g, Z_0} \underbrace{P(\mathbf{A}|Z_g, Z_0, \mathbf{C})}_{\text{逆动力学(规划)}} \underbrace{P(Z_g|Z_0, \mathbf{C})}_{\text{世界模型(想象)}} \underbrace{P(Z_0|\mathbf{C})}_{\text{状态估计(感知)}}
$$

- step1：引入潜空间进行全概率展开（边缘化 Marginalization）
    - $P(\mathbf{A}|\mathbf{C}) = \sum_{Z_g, Z_0} P(\mathbf{A}, Z_g, Z_0|\mathbf{C})$
- chain rule of probability
    - $P(\mathbf{A}, Z_g, Z_0|\mathbf{C}) = P(\mathbf{A}|Z_g, Z_0, \mathbf{C}) P(Z_g, Z_0|\mathbf{C})$
    - $P(Z_g, Z_0|\mathbf{C}) = P(Z_g|Z_0, \mathbf{C}) P(Z_0|\mathbf{C})$

### odds 与 bayes

> bayes filtering：基于有噪观测数据的统计推断（Statistical Inference）

- $O(H) = \frac{P(H)}{P(\neg H)} = \frac{P(H)}{1 - P(H)}$
    - 概率 $P(H)$：事件 $H$ 发生的可能性，范围是 $[0, 1]$。
    - 几率 $O(H)$：事件发生与不发生的比值。范围是 $[0, \infty)$。
- bayes
    - $P(H|E) = \frac{P(E|H) \cdot P(H)}{P(E)} \quad ......(1)$
    - $P(\neg H|E) = \frac{P(E|\neg H) \cdot P(\neg H)}{P(E)} \quad ......(2)$
    - 两式相除：$\frac{P(H|E)}{P(\neg H|E)} = \frac{P(E|H) \cdot P(H)}{P(E|\neg H) \cdot P(\neg H)}$

$$
\underbrace{\frac{P(H|E)}{P(\neg H|E)}}_{\text{后验几率}} = \underbrace{\frac{P(E|H)}{P(E|\neg H)}}_{\text{似然比}} \times \underbrace{\frac{P(H)}{P(\neg H)}}_{\text{先验几率}}
$$

- $\log O(H|E) = \log \Lambda(E|H) + \log O(H)$
    - Posterior Odds = Likelihood Ratio × Prior Odds。
